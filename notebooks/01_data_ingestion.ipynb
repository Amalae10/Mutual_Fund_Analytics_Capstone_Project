# 01 ? Data Ingestion

This notebook documents the ingestion of the raw mutual fund datasets used in the Bluestock Mutual Fund Analytics project.

## Objective

- Load raw CSV datasets from `data/raw/`
- Inspect dataset dimensions and data types
- Verify that the source datasets are available
- Run the project's ingestion script

```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path('../data/raw')

print('Raw data directory:', RAW_DIR.resolve())
print('\nAvailable datasets:')
for file in sorted(RAW_DIR.glob('*.csv')):
    print('-', file.name)
```

```python
fund_master = pd.read_csv(RAW_DIR / '01_fund_master.csv')
print('Fund master shape:', fund_master.shape)
display(fund_master.head())
print('\nData types:')
display(fund_master.dtypes)
```

```python
nav_history = pd.read_csv(RAW_DIR / '02_nav_history.csv')
print('NAV history shape:', nav_history.shape)
display(nav_history.head())
```
